```python
# 4 ( cc )

class Company:
    def __init__(self, name, age, salary):
        self.name = name
        self.age = age
        self.__salary = salary   # private variable
    def show(self):
        print("salary:", self.__salary)
class Employee(Company):
    def func(self):
        print("Employee name:", self.name)
        print("Employee age:", self.age)
        # private variable cannot be accessed directly here
emp = Employee("pranjal", 18, 250000)
print(emp.name)
emp.func()
emp.show()
```

```python
# private access

class Company:
    def __init__(self, name, age, salary):
        self.name = name          # public
        self._age = age           # protected
        self.__salary = salary    # private
    def show(self):
        print("salary:", self.__salary)
class Employee(Company):
    def func(self):
        print(self._age)  # allowed (protected)
emp = Employee("pranjal", 18, 25000)
print(emp.name)   # public
emp.func()        # protected
emp.show()        # private via method
```

```python
# 6 Encapsulation

class BankAccount:
    def __init__(self, acc_balance):
        self.__acc_balance = acc_balance

    def deposit(self, amount):
        self.__acc_balance += amount
        print("Balance:", self.__acc_balance)


user = BankAccount(3200)
user.deposit(1800)
```

```python
# Single Inheritence

class Information:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

    def print_info(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Gender:", self.gender)


# Child class
class Learner(Information):
    def __init__(self, name, age, gender, qual):
        super().__init__(name, age, gender)  # call parent constructor
        self.qual = qual

    def display(self):
        self.print_info()
        print("Qualification:", self.qual)


L1 = Learner("Pranjal", 18, "Male", "BCA")
L1.display()
```

```python
  # Parent class 1
class Father:
    father_name = "Akhilesh"
# Parent class 2
class Mother:
    mother_name = "Ritu"
# Child class (inherits from both)
class Child(Father, Mother):
    def parent(self):
        print(f"Father name is {self.father_name} and Mother name is {self.mother_name}")
# Object creation
c1 = Child()
c1.parent()
```
